# 📖 Rapport École du Sabbat — Colab (Python + R)

Tsindrio ny ▶ isaky ny cell araka ny filaharana.

| Cell | Asa |
|---|---|
| 1 | Fametrahana |
| 2 | **Ajouter classe** (hatramin'ny 10) |
| 3 | Angona isaky ny **Sabata** (isaky ny classe) |
| 4 | **Rapport isaky ny classe** |
| 5 | **Rapport totaly** ny classe rehetra |
| 6 | **Top 3**: Tonga tsy tara (%), Asa fitorina, Fanatitra |
| 7 | **Asa fitorina** (fitambaran'ireo asa 6) |
| 8 | Sary Python |
| 9 | Sary **R** (ggplot2) |
| 10 | Fanondranana **PDF** na **PowerPoint** |

**Asa fitorina** = Fampianarana Baiboly natao + Boky/trakta + Seminera + Ala Sarona + Nahavita fampianarana an-taratasy + Batisa.
**Tonga tsy tara %** = Tonga tsy tara ÷ (Tonga tsy tara + Tara) × 100.

```python
# ▶ CELL 1 — Fametrahana
!pip -q install python-pptx
import pandas as pd, numpy as np, os, matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
%load_ext rpy2.ipython
pd.set_option("display.width", 220); pd.set_option("display.max_columns", 40)

# Tsanganana ampidirina isaky ny classe isaky ny Sabata
FIELDS = {
 "tonga_tsy_tara":       "Membres tonga TSY TARA",
 "tonga_tara":           "Membres tonga TARA",
 "mpamangy":             "Mpamangy",
 "asa_soa":              "Asa soa natao",
 "fampianarana_baiboly": "Fampianarana Baiboly natao",
 "boky_trakta":          "Boky na trakta nozaraina",
 "seminera":             "Seminera na Kaonferansa natao",
 "ala_sarona":           "Fampianarana Baiboly an-taratasy ALA SARONA",
 "ala_sarona_vita":      "Nahavita fampianarana an-taratasy",
 "batisa":               "Batisa vokatry ny Tarika Miasa",
 "fanatitra_mission":    "Fanatitra ho an'ny asa maneran-tany",
}
# ASA FITORINA = fitambaran'ireo 6 ireo ihany (tsy ao ny Asa soa)
ASA_FITORINA = ["fampianarana_baiboly","boky_trakta","seminera",
                "ala_sarona","ala_sarona_vita","batisa"]
SHORT = {"mambra_tonga":"Membres tonga","tonga_tsy_tara":"Tsy tara","tonga_tara":"Tara",
 "tsy_tara_%":"Tsy tara %","mpamangy":"Mpamangy","isan_tonga":"Isan'ny tonga","asa_soa":"Asa soa",
 "fampianarana_baiboly":"Baiboly","boky_trakta":"Boky/trakta","seminera":"Seminera",
 "ala_sarona":"Ala Sarona","ala_sarona_vita":"Nahavita","batisa":"Batisa",
 "fanatitra_mission":"Fanatitra","asa_fitorina":"ASA FITORINA","daty":"Daty","classe":"Classe"}

DATA_FILE = "sabata.csv"
CLASSES = {}
df = pd.DataFrame(columns=["daty","classe"] + list(FIELDS))
print("✅ Vonona. Mandehana amin'ny Cell 2.")

# DEMO (hitsapana ihany)
DEMO = """
rng = np.random.default_rng(7)
CLASSES.clear(); df = df.iloc[0:0]
for i in range(1, 6): ampiditra_classe(f"Classe {i}", f"Mpampianatra {i}", 20 + i*3)
for d in pd.date_range("2026-10-03", periods=5, freq="7D"):
    for c in CLASSES:
        ts = int(rng.integers(8, 25))
        ampiditra_sabata(d, c, tonga_tsy_tara=ts, tonga_tara=int(rng.integers(0, 8)),
            mpamangy=int(rng.integers(0,6)), asa_soa=int(rng.integers(0,5)),
            fampianarana_baiboly=int(rng.integers(0,6)), boky_trakta=int(rng.integers(0,40)),
            seminera=int(rng.integers(0,2)), ala_sarona=int(rng.integers(0,4)),
            ala_sarona_vita=int(rng.integers(0,3)), batisa=int(rng.integers(0,2)),
            fanatitra_mission=float(rng.integers(5000,60000)))
"""
```

```python
# ▶ CELL 2 — AJOUTER CLASSE (hatramin'ny 10)
def ampiditra_classe(anarana, mpampianatra="", mambra=0):
    anarana = anarana.strip()
    if not anarana: return print("⚠️ Soraty ny anaran'ny classe.")
    if anarana in CLASSES: return print(f"⚠️ {anarana} efa misy.")
    if len(CLASSES) >= 10: return print("⚠️ Classe 10 no betsaka indrindra.")
    CLASSES[anarana] = {"mpampianatra": mpampianatra, "mambra": int(mambra)}
    print(f"✅ Nampiana: {anarana}  ({len(CLASSES)}/10)")

#@title Ampio classe vaovao { run: "auto" }
anarana_classe = "Classe 1" #@param {type:"string"}
mpampianatra   = "" #@param {type:"string"}
mambra_voasoratra = 0 #@param {type:"integer"}
ampiditra_classe(anarana_classe, mpampianatra, mambra_voasoratra)
print(pd.DataFrame(CLASSES).T)
```

```python
# ▶ CELL 3 — ANGONA ISAKY NY SABATA (isaky ny classe)
def ampiditra_sabata(daty, classe, **val):
    global df
    if classe not in CLASSES: return print(f"⚠️ {classe} tsy ao amin'ny classe. Ampio aloha (Cell 2).")
    row = {"daty": pd.to_datetime(daty).strftime("%Y-%m-%d"), "classe": classe}
    row.update({k: float(val.get(k, 0)) if k == "fanatitra_mission" else int(val.get(k, 0)) for k in FIELDS})
    df = df[~((df.daty == row["daty"]) & (df.classe == classe))]   # soloina raha efa misy
    df = pd.concat([df, pd.DataFrame([row])], ignore_index=True)
    df.to_csv(DATA_FILE, index=False)
    print(f"✅ Voatahiry: {classe} · {row['daty']}")

#@title Fampidirana isaky ny Sabata { run: false }
daty = "2026-11-07" #@param {type:"date"}
classe = "Classe 1" #@param {type:"string"}
tonga_tsy_tara = 0 #@param {type:"integer"}
tonga_tara = 0 #@param {type:"integer"}
mpamangy = 0 #@param {type:"integer"}
asa_soa = 0 #@param {type:"integer"}
fampianarana_baiboly = 0 #@param {type:"integer"}
boky_trakta = 0 #@param {type:"integer"}
seminera = 0 #@param {type:"integer"}
ala_sarona = 0 #@param {type:"integer"}
ala_sarona_vita = 0 #@param {type:"integer"}
batisa = 0 #@param {type:"integer"}
fanatitra_mission = 0.0 #@param {type:"number"}

ampiditra_sabata(daty, classe, tonga_tsy_tara=tonga_tsy_tara, tonga_tara=tonga_tara,
  mpamangy=mpamangy, asa_soa=asa_soa, fampianarana_baiboly=fampianarana_baiboly,
  boky_trakta=boky_trakta, seminera=seminera, ala_sarona=ala_sarona,
  ala_sarona_vita=ala_sarona_vita, batisa=batisa, fanatitra_mission=fanatitra_mission)

# 💡 Hitsapana amin'ny angona demo? Esory ny "#" eto ambany:
# exec(DEMO)
```

```python
# ▶ CELL 4 — RAPPORT ISAKY NY CLASSE
def _calc(d):
    d = d.copy()
    d["mambra_tonga"] = d.tonga_tsy_tara + d.tonga_tara                # Membres tonga = tsy tara + tara
    d["isan_tonga"]   = d.mambra_tonga + d.mpamangy                    # Isan'ny tonga
    d["asa_fitorina"] = d[ASA_FITORINA].sum(axis=1)                    # ASA FITORINA (6 asa)
    d["tsy_tara_%"]   = (100 * d.tonga_tsy_tara / d.mambra_tonga.replace(0, np.nan)).round(1)
    return d

def sivana(de=None, ha=None):
    d = df.copy()
    for k in FIELDS: d[k] = pd.to_numeric(d[k])
    if de: d = d[d.daty >= de]
    if ha: d = d[d.daty <= ha]
    return _calc(d)

COLS = ["tonga_tsy_tara","tonga_tara","mambra_tonga","tsy_tara_%","mpamangy","isan_tonga","asa_soa",
        "fampianarana_baiboly","boky_trakta","seminera","ala_sarona","ala_sarona_vita","batisa",
        "asa_fitorina","fanatitra_mission"]

def tatitra_classe(classe, de=None, ha=None):
    d = sivana(de, ha); d = d[d.classe == classe].sort_values("daty")
    if d.empty: return print("Tsy misy angona.")
    t = d.set_index("daty")[COLS]
    tot = t.drop(columns="tsy_tara_%").sum()
    tot["tsy_tara_%"] = round(100 * tot.tonga_tsy_tara / max(tot.mambra_tonga, 1), 1)
    t.loc["TOTALY"] = tot[COLS]
    return t

#@title Safidio ny classe { run: false }
classe_hojerena = "Classe 1" #@param {type:"string"}
date_de = "" #@param {type:"string"}
date_ha = "" #@param {type:"string"}
tatitra_classe(classe_hojerena, date_de or None, date_ha or None)
```

```python
# ▶ CELL 5 — RAPPORT TOTALY (classe rehetra)
def fitambarana(de=None, ha=None):
    d = sivana(de, ha)
    g = d.groupby("classe")[list(FIELDS) + ["mambra_tonga","isan_tonga","asa_fitorina"]].sum()
    g["tsy_tara_%"] = (100 * g.tonga_tsy_tara / g.mambra_tonga.replace(0, np.nan)).round(1)
    g.loc["TOTALY"] = g.sum(numeric_only=True)
    g.loc["TOTALY","tsy_tara_%"] = round(100*g.loc["TOTALY","tonga_tsy_tara"]/max(g.loc["TOTALY","mambra_tonga"],1),1)
    return g[COLS]

agg = fitambarana()
agg
```

```python
# ▶ CELL 6 — TOP 3 (classe 3 voalohany)
def top3(col, de=None, ha=None, cols=None):
    g = fitambarana(de, ha).drop("TOTALY")
    t = g.sort_values(col, ascending=False).head(3)
    t = t[cols or [col]].copy(); t.insert(0, "Laharana", range(1, len(t)+1))
    return t

TOP_TSY_TARA = ["tonga_tsy_tara","tonga_tara","mambra_tonga","tsy_tara_%"]
TOP_ASA      = ASA_FITORINA + ["asa_fitorina"]

print("🏆 TOP 3 — TONGA TSY TARA (ambony indrindra amin'ny %)"); display(top3("tsy_tara_%", cols=TOP_TSY_TARA))
print("🏆 TOP 3 — ASA FITORINA (fitambaran'ny asa 6)");           display(top3("asa_fitorina", cols=TOP_ASA))
print("🏆 TOP 3 — FANATITRA ho an'ny asa maneran-tany");          display(top3("fanatitra_mission"))
```

```python
# ▶ CELL 7 — ASA FITORINA isaky ny classe (tena nanao asa fitorina)
asa = fitambarana().drop("TOTALY")[TOP_ASA].sort_values("asa_fitorina", ascending=False)
asa.loc["TOTALY"] = asa.sum()
asa.rename(columns=SHORT)
```

```python
# ▶ CELL 8 — SARY (Python)
def sary_top3(path="sary_top3.png", de=None, ha=None):
    g = fitambarana(de, ha).drop("TOTALY")
    fig, ax = plt.subplots(1, 3, figsize=(15, 4.6))
    for a, (col, ttl, clr, suf) in zip(ax, [("tsy_tara_%","Top 3 Tonga tsy tara (%)","#1e5a8a"," %"),
                                            ("asa_fitorina","Top 3 Asa fitorina","#2f855a",""),
                                            ("fanatitra_mission","Top 3 Fanatitra","#c28a2b","")]):
        s = g[col].sort_values(ascending=False).head(3)
        a.bar(s.index, s.values, color=clr); a.set_title(ttl)
        for i, v in enumerate(s.values): a.text(i, v, f"{v:,.0f}{suf}", ha="center", va="bottom")
        a.tick_params(axis="x", rotation=20)
    plt.tight_layout(); plt.savefig(path, dpi=150); plt.show()

sary_top3()
```

```python
# ▶ CELL 10 — FANONDRANANA: PDF na POWERPOINT
from pptx import Presentation
from pptx.util import Inches, Pt
from pptx.dml.color import RGBColor

def _f(v):
    if isinstance(v, (int, float, np.integer, np.floating)) and not pd.isna(v):
        return f"{v:,.1f}" if (isinstance(v, (float, np.floating)) and v != int(v)) else f"{int(v):,}"
    return "" if pd.isna(v) else str(v)

def _pages(de=None, ha=None):
    """Lisitry ny pejy: (lohateny, DataFrame na sary)"""
    g = fitambarana(de, ha)
    P = []
    P.append(("Rapport totaly — classe rehetra", g[["tonga_tsy_tara","tonga_tara","mambra_tonga","tsy_tara_%","mpamangy",
              "isan_tonga","asa_soa","asa_fitorina","fanatitra_mission"]].rename(columns=SHORT).rename_axis("Classe")))
    sary_top3("_top3.png", de, ha); plt.close("all")
    P.append(("Top 3", "_top3.png"))
    P.append(("Top 3 — Tonga tsy tara (%)", top3("tsy_tara_%", de, ha, TOP_TSY_TARA).rename(columns=SHORT)))
    a = top3("asa_fitorina", de, ha, TOP_ASA).rename(columns=SHORT)
    P.append(("Top 3 — Asa fitorina (fitambaran'ny asa 6)", a))
    P.append(("Top 3 — Fanatitra ho an'ny asa maneran-tany", top3("fanatitra_mission", de, ha).rename(columns=SHORT)))
    for c in CLASSES:
        t = tatitra_classe(c, de, ha)
        if t is not None:
            P.append((f"{c} — {CLASSES[c]['mpampianatra']}", t.rename(columns=SHORT).rename_axis("Daty")))
    return P

def _pdf_table(pdf, d, titre):
    d = d.reset_index()
    fig, ax = plt.subplots(figsize=(13.33, 7.5)); ax.axis("off")
    ax.set_title(titre, fontsize=20, loc="left", color="#1e5a8a", pad=20)
    cells = [[_f(v) for v in r] for r in d.values]
    h = min(0.075 * (len(cells) + 1), 0.9)
    t = ax.table(cellText=cells, colLabels=[str(c) for c in d.columns], bbox=[0, 1 - h, 1, h], cellLoc="center")
    t.auto_set_font_size(False); t.set_fontsize(8 if d.shape[1] > 10 else 11)
    for (r, c), cell in t.get_celld().items():
        if r == 0: cell.set_facecolor("#1e5a8a"); cell.set_text_props(color="white", weight="bold", fontsize=(6.5 if d.shape[1] > 10 else 10))
        elif str(d.iloc[r-1, 0]) == "TOTALY": cell.set_facecolor("#f3ead6"); cell.set_text_props(weight="bold")
    pdf.savefig(fig); plt.close(fig)

def _pdf(nom, lohateny, P):
    with PdfPages(nom) as pdf:
        fig = plt.figure(figsize=(13.33, 7.5)); fig.patch.set_facecolor("#1e5a8a")
        fig.text(.5, .55, lohateny, ha="center", color="white", fontsize=36, weight="bold")
        fig.text(.5, .45, f"{df.daty.min()}  →  {df.daty.max()}", ha="center", color="#e5e1d8", fontsize=18)
        pdf.savefig(fig); plt.close(fig)
        for titre, x in P:
            if isinstance(x, str):
                fig, ax = plt.subplots(figsize=(13.33, 7.5)); ax.axis("off")
                ax.set_title(titre, fontsize=20, loc="left", color="#1e5a8a"); ax.imshow(plt.imread(x))
                pdf.savefig(fig); plt.close(fig)
            else: _pdf_table(pdf, x, titre)

def _pptx(nom, lohateny, P):
    prs = Presentation(); prs.slide_width, prs.slide_height = Inches(13.33), Inches(7.5)
    s = prs.slides.add_slide(prs.slide_layouts[0]); s.shapes.title.text = lohateny
    s.placeholders[1].text = f"{df.daty.min()} → {df.daty.max()}"
    for titre, x in P:
        s = prs.slides.add_slide(prs.slide_layouts[5]); s.shapes.title.text = titre
        s.shapes.title.text_frame.paragraphs[0].font.size = Pt(28)
        if isinstance(x, str):
            s.shapes.add_picture(x, Inches(0.5), Inches(1.7), width=Inches(12.3)); continue
        d = x.reset_index(); n, m = d.shape[0] + 1, d.shape[1]
        t = s.shapes.add_table(n, m, Inches(0.3), Inches(1.5), Inches(12.7), Inches(0.38*n)).table
        for j, c in enumerate(d.columns): t.cell(0, j).text = str(c)
        for i, r in enumerate(d.values):
            for j, v in enumerate(r): t.cell(i+1, j).text = _f(v)
        for row in t.rows:
            for cell in row.cells:
                for p in cell.text_frame.paragraphs: p.font.size = Pt(10 if m > 10 else 13)
    prs.save(nom)

def ataovy_rapport(format="PDF", lohateny="Rapport École du Sabbat", de=None, ha=None):
    if df.empty: return print("⚠️ Tsy misy angona.")
    P = _pages(de, ha); sortie = []
    if format in ("PDF", "Roa"):        _pdf("Rapport_Ecole_du_Sabbat.pdf", lohateny, P);   sortie.append("Rapport_Ecole_du_Sabbat.pdf")
    if format in ("PowerPoint", "Roa"): _pptx("Rapport_Ecole_du_Sabbat.pptx", lohateny, P); sortie.append("Rapport_Ecole_du_Sabbat.pptx")
    print("✅", ", ".join(sortie))
    try:
        from google.colab import files
        for f in sortie: files.download(f)
    except Exception: pass

#@title Fanondranana { run: false }
format_rapport = "PDF" #@param ["PDF","PowerPoint","Roa"]
lohateny_rapport = "Rapport École du Sabbat" #@param {type:"string"}
daty_de = "" #@param {type:"string"}
daty_ha = "" #@param {type:"string"}
ataovy_rapport(format_rapport, lohateny_rapport, daty_de or None, daty_ha or None)
```
